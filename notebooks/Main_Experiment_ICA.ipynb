## ICA Artifact Rejection
Fit ICA on the concatenated epochs to identify and remove artifactual components (eye blinks, muscle, etc.).

**Workflow:**
1. Run the cell below to fit ICA and inspect component plots.
2. Set `ica.exclude = [...]` with the component indices you want to remove.
3. Run the "Apply ICA" cell to get cleaned epochs.

```python
from mne.preprocessing import ICA

# Fit ICA on combined epochs
_n_components_epochs = min(25, len(epochs.ch_names) - 1)

# Build a continuous Raw from the epoch data
data = epochs.get_data()  # (n_epochs, n_ch, n_times)
data_concat = data.transpose(1, 0, 2).reshape(len(epochs.ch_names), -1)
raw_from_epochs = mne.io.RawArray(data_concat, epochs.info.copy())

print("Fitting ICA on epoch data...")
ica = ICA(
    n_components=_n_components_epochs,
    method="fastica",
    random_state=97
)
ica.fit(raw_from_epochs, decim=3)
print(f"ICA fitted ({ica.n_components_} components).")

# Plot components for visual inspection
ica.plot_components()
ica.plot_sources(raw_from_epochs, start=0, stop=60)

print(f"Current ica.exclude = {ica.exclude}")
```

```python
# Run ICLabel classification
from mne_icalabel import label_components
import pandas as pd

print("Running ICLabel classification...")
label_dict = label_components(raw_from_epochs, ica, method='iclabel')

ic_labels = label_dict['labels']
ic_probs = label_dict['y_pred_proba']

# Build results table
results = []
for idx, (label, prob) in enumerate(zip(ic_labels, ic_probs)):
    max_prob = prob.max() if hasattr(prob, '__len__') and len(prob) > 1 else prob
    results.append({'Component': idx, 'Label': label, 'Probability': f"{max_prob:.3f}"})

results_df = pd.DataFrame(results)
print("\n" + results_df.to_string(index=False))

# Suggest exclusions
exclude_labels = ['eye blink', 'eye movement', 'muscle artifact']
auto_exclude = [i for i, label in enumerate(ic_labels) if label in exclude_labels]
print(f"\nSuggested exclusions: {auto_exclude}")
```

## Channel Selection & CSD
Now reduce to the electrode group of interest and optionally apply CSD.

```python
# Pick electrode group
filtered_electrodes = [elec for elec in params_dict["Electorde_Group"] if elec in epochs.ch_names]
epochs.pick(filtered_electrodes)
print(f"Picked {len(filtered_electrodes)} channels: {filtered_electrodes[:5]}...")

# Optionally apply CSD
if params_dict.get('PerformCsd', False):
    print("Applying Current Source Density...")
    epochs = mne.preprocessing.compute_current_source_density(epochs)
    print("CSD applied.")

# Restore full rank after ICA (prevents SVD convergence errors in CSP)
n_channels = len(epochs.ch_names)
data_2d = epochs.get_data().transpose(1, 0, 2).reshape(n_channels, -1)
rank = np.linalg.matrix_rank(data_2d)
if rank < n_channels:
    from sklearn.decomposition import PCA
    print(f"Rank-deficient data ({rank}/{n_channels}), applying PCA to restore full rank...")
    pca = PCA(n_components=rank)
    data_3d = epochs.get_data()                         # (n_epochs, n_ch, n_times)
    n_epochs, _, n_times = data_3d.shape
    flat = data_3d.transpose(0, 2, 1).reshape(-1, n_channels)  # (n_epochs*n_times, n_ch)
    reduced = pca.fit_transform(flat)                   # (n_epochs*n_times, rank)
    data_reduced = reduced.reshape(n_epochs, n_times, rank).transpose(0, 2, 1)
    info_new = mne.create_info([f"PC{i}" for i in range(rank)], epochs.info['sfreq'], ch_types='eeg')
    epochs = mne.EpochsArray(data_reduced, info_new, events=epochs.events, tmin=epochs.tmin, event_id=epochs.event_id)
    print(f"Reduced to {rank} PCA components.")
else:
    print(f"Data is full rank ({rank}/{n_channels}), no reduction needed.")

epochs_filtered = epochs.filter(params_dict['LowPass'], params_dict['HighPass'], method=params_dict['filter_method'], iir_params = dict(order=4, ftype='butter'),phase='forward',pad=0)
epochs=epochs_filtered
```
